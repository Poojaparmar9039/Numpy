## Advance operation with business examples

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
sales_data=np.array([
    [1,15000,18000,220000,250000],
    [2,12000,14000,160000,190000],
    [3,20000,23000,260000,300000],
    [4,18000,21000,240000,270000],
    [5,24000,38500,205000,230000]

])
print(sales_data)
```

```python
print("shape:",sales_data.shape)
```

```python
print("data for Ist three sales \n",sales_data[:3])
```

```python
print("data of 2 nd restaurant \n",sales_data[2])
```

```python
print("data of 3 restaurant of 3 value",sales_data[2][2])
```

```python
print(sales_data[:,1:])
```

```python
print(np.sum(sales_data[1:3]))
print(np.sum(sales_data))
print(np.sum(sales_data,axis=0))
```

```python
yearly=np.sum(sales_data[:,:],axis=0)
print(yearly)
```

```python
min_sales=np.min(sales_data[:,1:],axis=0)
print(min_sales)
```

```python
min_sales=np.min(sales_data[:,1:],axis=1)
print(min_sales)
```

```python
max_sales=np.max(sales_data[:,1:],axis=0)
print(max_sales)
```

```python
avg_sales=np.mean(sales_data[:,1:],axis=1)
print(avg_sales)
```

```python
#cumulative sum----we add previous year sale to next year sum
cum_sum=np.cumsum(sales_data[:,1:],axis=1)
print(cum_sum)
```

```python
plt.figure(figsize=(10,6))
plt.plot(np.mean(cum_sum,axis=0))
plt.title("average cumulative sales across all restaurant")
plt.xlabel("years")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```

```python
vector1=np.array([1,2,3,4,5])
vector2=np.array([6,7,8,9,10])
print(vector1+vector2)
print(vector1*vector2)
print(vector1-vector2)
print(np.dot(vector1,vector2))
```
